```python
from pathlib import Path

md_files = list(Path("data/raw").rglob("*.md"))

len(md_files), md_files[:10]
```

```python
path = md_files[0]

print(path)
```

```python
text = path.read_text(encoding="utf-8")
print(text[:2000])
```

```python
from app.rag.chunking import chunk_markdown


chunks = chunk_markdown(text, max_chunk_size=1000)

print("Количество chunks:", len(chunks))
```

```python
for i, chunk in enumerate(chunks[:100]):
    print("=" * 80)
    print("CHUNK:", i)
    print("HEADING:", chunk["heading_path"])
    #print()
    #print(chunk["content"])
```

```python
sizes = [len(chunk["content"]) for chunk in chunks]

print("Минимальный:", min(sizes))
print("Максимальный:", max(sizes))
print("Средний:", sum(sizes) / len(sizes))
```

```python
oversized = [
    (i, len(chunk["content"]), chunk["heading_path"])
    for i, chunk in enumerate(chunks)
    if len(chunk["content"]) > 1000
]

print("Слишком больших chunks:", len(oversized))
print(oversized[:10])
```
